```python
from sklearn.cluster import KMeans
from sklearn.preprocessing import LabelEncoder, StandardScaler

le_bolum = LabelEncoder()
le_kitap = LabelEncoder()

veri["BolumKod"] = le_bolum.fit_transform(veri["Bolum"])
veri["KitapKod"] = le_kitap.fit_transform(veri["KitapAdi"])

X = veri[["BolumKod", "KitapKod"]]

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

kmeans = KMeans(n_clusters=5, random_state=42)

veri["Kume"] = kmeans.fit_predict(X_scaled)
```
